## Phân loại cột numerical và categorical

Cell này chia các cột thành 2 nhóm chính:
- `numerical columns`: các cột năm như `1960`, `1961`, ..., `2024`,
- `categorical columns`: các cột mô tả như `Country Name`, `Country Code`, `Indicator Name`, `Indicator Code`.

Việc phân loại này quan trọng vì:
- nhóm numerical sẽ được dùng cho thống kê mô tả, missing values và visualization,
- nhóm categorical sẽ được dùng để nhận diện quốc gia, mã nước và metadata của dataset.


```python
numerical_cols = [col for col in gdp_df.columns if str(col).isdigit()]
categorical_cols = [col for col in gdp_df.columns if col not in numerical_cols]

print("Numerical columns:", len(numerical_cols))
print(numerical_cols)

print("\nCategorical columns:", len(categorical_cols))
print(categorical_cols)
```

## Kiểm tra số lượng giá trị thiếu

Cell này tạo bảng tóm tắt missing values cho từng cột, bao gồm:
- `null_count`: số lượng giá trị thiếu,
- `null_ratio_pct`: tỷ lệ phần trăm giá trị thiếu,

Đây là bước quan trọng để đánh giá chất lượng dữ liệu.

## Phân biệt country rows và aggregate rows

Cell này dùng `Country Code` để tách:
- các dòng đại diện cho quốc gia hoặc lãnh thổ thật,
- các dòng aggregate như khu vực, nhóm thu nhập hoặc nhóm kinh tế.

Điều này rất quan trọng vì dataset WDI không chỉ chứa dữ liệu của từng nước, mà còn chứa nhiều dòng tổng hợp.  
Nếu không tách riêng, kết quả phân tích có thể bị sai lệch.


```python
SPECIAL_COUNTRY_CODES = {"XKX", "SXM"}

def is_country_code(code):
    code = str(code).strip()
    if code in SPECIAL_COUNTRY_CODES:
        return True
    try:
        alpha2 = pc.country_alpha3_to_country_alpha2(code)
        pc.country_alpha2_to_continent_code(alpha2)
        return True
    except:
        return False

country_mask = gdp_df["Country Code"].astype(str).apply(is_country_code)

print("Country / territory rows:", int(country_mask.sum()))
print("Aggregate rows:", int((~country_mask).sum()))
```

## Xem danh sách aggregate codes

Cell này liệt kê các mã không phải country thật, tức là các `aggregate codes`.

Mục đích:
- xác định rõ dataset đang chứa những nhóm tổng hợp nào,
- hỗ trợ giải thích tại sao số dòng của dataset lớn hơn số lượng quốc gia thực tế,
- làm cơ sở để loại bỏ aggregate rows khi chỉ muốn phân tích ở cấp quốc gia.


```python
aggregate_codes = gdp_df.loc[~country_mask, "Country Code"].astype(str).unique().tolist()

print("Aggregate codes:")
print(aggregate_codes)
```

## Kiểm tra độ đầy đủ dữ liệu theo từng năm

Cell này tính số lượng dòng có dữ liệu ở mỗi cột năm.

Output cho biết:
- năm nào có coverage cao,
- năm nào bị thiếu dữ liệu,
- tỷ lệ hoàn chỉnh dữ liệu theo từng năm.

Đây là bước quan trọng để xác định mức độ đầy đủ dữ liệu của GDP qua thời gian.


```python
year_non_null = gdp_df[numerical_cols].notna().sum().sort_index()

coverage_df = pd.DataFrame({
    "Year": year_non_null.index,
    "non_null_count": year_non_null.values,
    "non_null_ratio_pct": (year_non_null.values / len(gdp_df) * 100).round(2)
})

coverage_df.tail(50)
```

```python
latest_year_with_data = coverage_df.loc[coverage_df["non_null_count"] > 0, "Year"].iloc[-1]

print("Latest year with any data:", latest_year_with_data)
print("Non-null rows in latest year:", int(gdp_df[latest_year_with_data].notna().sum()))
```

## Tổng hợp các data issues của GDP dataset

Cell này tổng hợp các vấn đề chính của dataset dưới dạng nhận xét tự động.

Các issue được kiểm tra gồm:
- có cột rác hay không,
- có aggregate rows hay không,
- dataset có chứa đúng 1 indicator duy nhất không,
- có duplicate rows không,
- có cột nào null 100% không,
- có năm nào missing quá cao không,
- cột năm mới nhất có đủ dữ liệu không,
- phân phối dữ liệu có bị lệch mạnh hay không,
- dữ liệu đang ở wide format hay long format.



```python
issues = []

if unnamed_cols:
    issues.append(f"Co cot rac/technical column: {unnamed_cols}")

if (~country_mask).sum() > 0:
    issues.append(
        f"Dataset chua ca aggregate rows, khong chi country rows: {(~country_mask).sum()} dong aggregate, {country_mask.sum()} dong country/territory."
    )

if gdp_df["Indicator Name"].nunique() == 1 and gdp_df["Indicator Code"].nunique() == 1:
    issues.append("File chi chua 1 indicator duy nhat la GDP per capita (current US$).")

if gdp_df.duplicated().sum() == 0:
    issues.append("Khong co duplicate row hoan toan trung nhau.")
else:
    issues.append(f"Co {gdp_df.duplicated().sum()} duplicate rows.")

all_null_cols = gdp_null.loc[gdp_null["null_count"] == len(gdp_df), "column"].tolist()
if all_null_cols:
    issues.append(f"Co cot nam bi null 100%: {all_null_cols}")
else:
    issues.append("Khong co cot nao bi null 100%.")

high_missing_cols = gdp_null.loc[
    (gdp_null["column"].isin(numerical_cols)) & (gdp_null["null_ratio_pct"] >= 30),
    "column"
].tolist()

if high_missing_cols:
    issues.append(f"Cac nam co missing rat cao (>=30%): {high_missing_cols}")
else:
    issues.append("Khong co cot nam nao missing tu 30% tro len.")

count_2024 = int(gdp_df["2024"].notna().sum())
issues.append(f"Cot 2024 chua day du, hien co {count_2024}/{len(gdp_df)} dong co du lieu.")

latest_series = pd.to_numeric(gdp_df[latest_year_with_data], errors="coerce").dropna()
mean_val = latest_series.mean()
median_val = latest_series.median()

if median_val != 0 and mean_val / median_val >= 2:
    issues.append(
        f"Phan phoi GDP lech phai rat manh/outlier cao vi mean ({mean_val:,.2f}) lon hon median ({median_val:,.2f}) rat nhieu."
    )

issues.append("Du lieu dang o wide format theo nam, can melt sang long format neu muon phan tich time series hoac merge voi dataset khac.")

for i, issue in enumerate(issues, start=1):
    print(f"{i}. {issue}")
```
