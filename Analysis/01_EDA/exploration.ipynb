```python
# ======================================
# 1. IMPORT LIBRARIES
# ======================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pycountry_convert as pc

sns.set_style("whitegrid")
sns.set_palette("Set2")


# ======================================
# 2. LOAD & CLEAN DATA (FIX ALL ISSUES)
# ======================================
def load_clean(path):
    df = pd.read_csv(
        path,
        skiprows=2,              # bỏ 2 dòng metadata
        sep=None,                # auto detect delimiter
        engine='python',
        encoding='latin1',
        on_bad_lines='skip'
    )
    
    # clean column names
    df.columns = df.columns.str.strip()
    
    return df


gdp = load_clean("gdp.csv")
life = load_clean("life_expectancy.csv")
pop = load_clean("population.csv")


# ======================================
# 3. DEBUG (CHECK COLUMNS)
# ======================================
print("GDP columns:", gdp.columns)


# ======================================
# 4. BASIC STATS
# ======================================
def basic_stats(df, name):
    print(f"\n===== {name} =====")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

basic_stats(gdp, "GDP")
basic_stats(life, "Life Expectancy")
basic_stats(pop, "Population")


# ======================================
# 5. ADD CONTINENT
# ======================================
def country_to_continent(country):
    try:
        code = pc.country_name_to_country_alpha2(country)
        return pc.country_alpha2_to_continent_code(code)
    except:
        return None

gdp["continent"] = gdp["Country Name"].apply(country_to_continent)
life["continent"] = life["Country Name"].apply(country_to_continent)
pop["continent"] = pop["Country Name"].apply(country_to_continent)


# ======================================
# 6. CONVERT TO LONG FORMAT
# ======================================
def to_long(df, value_name):
    df_long = df.melt(
        id_vars=["Country Name", "Country Code", "continent"],
        var_name="Year",
        value_name=value_name
    )
    
    df_long["Year"] = pd.to_numeric(df_long["Year"], errors='coerce')
    df_long[value_name] = pd.to_numeric(df_long[value_name], errors='coerce')
    
    return df_long.dropna()

gdp_long = to_long(gdp, "GDP")
life_long = to_long(life, "Life")
pop_long = to_long(pop, "Population")


# ======================================
# 7. ADD DECADE
# ======================================
def add_decade(df):
    df["Decade"] = (df["Year"] // 10) * 10
    return df

gdp_long = add_decade(gdp_long)
life_long = add_decade(life_long)
pop_long = add_decade(pop_long)


# ======================================
# 8. GLOBAL TREND
# ======================================
def plot_trend(df, col, title):
    trend = df.groupby("Year")[col].mean()
    
    plt.figure(figsize=(10,6))
    plt.plot(trend.index, trend.values)
    plt.title(title)
    plt.xlabel("Year")
    plt.ylabel(col)
    plt.show()

plot_trend(gdp_long, "GDP", "Global GDP Trend")
plot_trend(life_long, "Life", "Global Life Expectancy Trend")
plot_trend(pop_long, "Population", "Global Population Trend")


# ======================================
# 9. TREND BY CONTINENT
# ======================================
def trend_by_continent(df, col):
    data = df.groupby(["Year", "continent"])[col].mean().reset_index()
    
    plt.figure(figsize=(10,6))
    sns.lineplot(data=data, x="Year", y=col, hue="continent")
    plt.title(f"{col} by Continent")
    plt.show()

trend_by_continent(gdp_long, "GDP")
trend_by_continent(life_long, "Life")
trend_by_continent(pop_long, "Population")


# ======================================
# 10. DISTRIBUTION BY DECADE
# ======================================
def plot_decade_distribution(df, col, title):
    plt.figure(figsize=(12,6))
    sns.boxplot(x="Decade", y=col, data=df)
    plt.title(title)
    plt.xticks(rotation=45)
    plt.show()

plot_decade_distribution(gdp_long, "GDP", "GDP Distribution by Decade")
plot_decade_distribution(life_long, "Life", "Life Expectancy by Decade")
plot_decade_distribution(pop_long, "Population", "Population by Decade")


# ======================================
# 11. HEATMAP
# ======================================
def plot_heatmap(df, col, title):
    pivot = df.groupby(["continent","Decade"])[col].mean().unstack()
    
    plt.figure(figsize=(10,6))
    sns.heatmap(pivot, annot=True, fmt=".0f", cmap="coolwarm")
    plt.title(title)
    plt.show()

plot_heatmap(gdp_long, "GDP", "GDP Heatmap")
plot_heatmap(life_long, "Life", "Life Expectancy Heatmap")
plot_heatmap(pop_long, "Population", "Population Heatmap")


# ======================================
# 12. LOG TREND
# ======================================
def plot_log_trend(df, col, title):
    trend = df.groupby("Year")[col].mean()
    
    plt.figure(figsize=(10,6))
    plt.plot(trend.index, np.log1p(trend.values))
    plt.title(title + " (Log Scale)")
    plt.show()

plot_log_trend(gdp_long, "GDP", "GDP Trend")
plot_log_trend(pop_long, "Population", "Population Trend")


# ======================================
# 13. MERGE DATA FOR ANIMATION
# ======================================
df_merged = gdp_long.merge(
    life_long, on=["Country Name","Country Code","Year","continent"]
).merge(
    pop_long, on=["Country Name","Country Code","Year","continent"]
)


# ======================================
# 14. ANIMATED BUBBLE CHART
# ======================================
fig = px.scatter(
    df_merged,
    x="GDP",
    y="Life",
    size="Population",
    color="continent",
    hover_name="Country Name",
    animation_frame="Year",
    log_x=True,
    size_max=60,
    title="GDP vs Life Expectancy Over Time"
)

fig.show()


# ======================================
# 15. GEO MAP (GDP)
# ======================================
fig = px.choropleth(
    gdp_long,
    locations="Country Name",
    locationmode="country names",
    color="GDP",
    hover_name="Country Name",
    animation_frame="Year",
    color_continuous_scale="Viridis",
    title="Global GDP Distribution Over Time"
)

fig.show()


# ======================================
# 16. GEO MAP (LIFE)
# ======================================
fig = px.choropleth(
    life_long,
    locations="Country Name",
    locationmode="country names",
    color="Life",
    hover_name="Country Name",
    animation_frame="Year",
    color_continuous_scale="Plasma",
    title="Life Expectancy Across the World"
)

fig.show()


# ======================================
# 17. GEO MAP (POPULATION)
# ======================================
fig = px.choropleth(
    pop_long,
    locations="Country Name",
    locationmode="country names",
    color="Population",
    hover_name="Country Name",
    animation_frame="Year",
    color_continuous_scale="Turbo",
    title="Population Growth Worldwide"
)

fig.show()
```

```python
import json
import warnings
from urllib.request import urlopen

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pycountry_convert as pc

from IPython.display import display, Markdown

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)
pd.set_option("display.max_columns", 100)
```

```python
GDP_PATH = "gdp.csv"
LIFE_PATH = "life_expectancy.csv"
POP_PATH = "population.csv"

CONTINENT_MAP = {
    "AF": "Africa",
    "AS": "Asia",
    "EU": "Europe",
    "NA": "North America",
    "OC": "Oceania",
    "SA": "South America",
    "AN": "Antarctica",
}

SPECIAL_CONTINENT = {
    "XKX": "Europe",        # Kosovo
    "SXM": "North America", # Sint Maarten
}

def country_to_continent(alpha3, alpha2=None):
    alpha3 = str(alpha3).strip()
    if alpha3 in SPECIAL_CONTINENT:
        return SPECIAL_CONTINENT[alpha3]

    try:
        if pd.isna(alpha2) or str(alpha2).strip() == "":
            alpha2 = pc.country_alpha3_to_country_alpha2(alpha3)
        continent_code = pc.country_alpha2_to_continent_code(str(alpha2).strip())
        return CONTINENT_MAP.get(continent_code)
    except:
        return None

def fetch_world_bank_metadata():
    url = "https://api.worldbank.org/v2/country?format=json&per_page=400"
    with urlopen(url) as response:
        payload = json.load(response)

    rows = []
    for item in payload[1]:
        rows.append({
            "Country Code": item.get("id"),
            "ISO2 Code": item.get("iso2Code"),
            "WB Country Name": item.get("name"),
            "WB Region": (item.get("region") or {}).get("value", ""),
            "Admin Region": (item.get("adminregion") or {}).get("value", ""),
            "Income Level": (item.get("incomeLevel") or {}).get("value", ""),
        })

    meta = pd.DataFrame(rows)
    meta["WB Region"] = meta["WB Region"].fillna("").str.strip().replace("", "Aggregates")
    meta["Is Country"] = meta["WB Region"].ne("Aggregates")
    meta["Continent"] = meta.apply(
        lambda row: country_to_continent(row["Country Code"], row["ISO2 Code"]),
        axis=1
    )
    return meta

metadata = fetch_world_bank_metadata()
metadata.head()
```

```python
def load_wdi(path):
    df = pd.read_csv(path, skiprows=4)
    df = df.drop(columns=["Unnamed: 69"], errors="ignore")

    year_cols = [c for c in df.columns if str(c).isdigit()]
    df[year_cols] = df[year_cols].apply(pd.to_numeric, errors="coerce")
    return df

def to_long(df, value_name):
    year_cols = [c for c in df.columns if str(c).isdigit()]
    long_df = df.melt(
        id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
        value_vars=year_cols,
        var_name="Year",
        value_name=value_name
    )
    long_df["Year"] = pd.to_numeric(long_df["Year"], errors="coerce").astype("Int64")
    long_df[value_name] = pd.to_numeric(long_df[value_name], errors="coerce")
    return long_df

def run_exploration(path, value_name, title, log_scale=False):
    raw_df = load_wdi(path)
    long_df = to_long(raw_df, value_name)

    df = long_df.merge(
        metadata[["Country Code", "WB Region", "Admin Region", "Continent", "Is Country"]],
        on="Country Code",
        how="left"
    )

    year_cols = [c for c in raw_df.columns if str(c).isdigit()]
    latest_year = int(df.loc[df["Is Country"] & df[value_name].notna(), "Year"].max())
    latest_df = df[(df["Year"] == latest_year) & (df["Is Country"])].copy()

    summary_df = pd.DataFrame({
        "Metric": [
            "So dong",
            "So cot",
            "So cot nam",
            "So dong country/territory",
            "So dong aggregate",
            "Tong missing o cac cot nam",
            "Ty le missing o cac cot nam",
            "Nam moi nhat co du lieu cap quoc gia",
            "So nuoc co du lieu o nam moi nhat",
        ],
        "Value": [
            raw_df.shape[0],
            raw_df.shape[1],
            len(year_cols),
            int(metadata["Is Country"].sum()),
            int((~metadata["Is Country"]).sum()),
            int(raw_df[year_cols].isna().sum().sum()),
            round(raw_df[year_cols].isna().sum().sum() / raw_df[year_cols].size, 4),
            latest_year,
            int(latest_df[value_name].notna().sum()),
        ]
    })

    desc_df = latest_df[value_name].dropna().describe(
        percentiles=[0.1, 0.25, 0.5, 0.75, 0.9]
    ).to_frame(name="Value")

    continent_df = (
        latest_df.dropna(subset=["Continent", value_name])
        .groupby("Continent")[value_name]
        .agg(["count", "mean", "median", "min", "max"])
        .sort_values("count", ascending=False)
        .reset_index()
    )

    region_df = (
        latest_df.dropna(subset=["WB Region", value_name])
        .groupby("WB Region")[value_name]
        .agg(["count", "mean", "median", "min", "max"])
        .sort_values("count", ascending=False)
        .reset_index()
    )

    coverage_df = (
        df[df["Is Country"]]
        .groupby("Year")[value_name]
        .agg(non_null_count="count")
        .reset_index()
    )

    display(Markdown(f"# {title}"))
    display(Markdown(f"**Latest year with country-level data:** `{latest_year}`"))

    print("1. Basic statistics")
    display(summary_df)

    print("2. Distribution statistics at latest year")
    display(desc_df)

    print("3. Group by continent")
    display(continent_df)

    print("4. Group by region")
    display(region_df)

    print("5. Data coverage by year")
    plt.figure(figsize=(13, 4))
    sns.lineplot(data=coverage_df, x="Year", y="non_null_count", marker="o")
    plt.title(f"{title} - so nuoc co du lieu theo nam")
    plt.xlabel("Year")
    plt.ylabel("Country count")
    plt.show()

    print("6. Distribution plot")
    plot_series = latest_df[value_name].dropna()
    if log_scale:
        plot_series = plot_series[plot_series > 0]

    fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
    sns.histplot(plot_series, bins=30, kde=True, ax=axes[0], color="#2a9d8f")
    sns.boxplot(x=plot_series, ax=axes[1], color="#e9c46a")

    if log_scale:
        axes[0].set_xscale("log")
        axes[1].set_xscale("log")
        axes[0].set_xlabel(f"{value_name} (log scale)")
        axes[1].set_xlabel(f"{value_name} (log scale)")
    else:
        axes[0].set_xlabel(value_name)
        axes[1].set_xlabel(value_name)

    axes[0].set_title(f"{title} - histogram")
    axes[1].set_title(f"{title} - boxplot")
    plt.tight_layout()
    plt.show()

    print("7. Geo map by continent")
    fig1 = px.choropleth(
        latest_df.dropna(subset=["Continent"]),
        locations="Country Code",
        locationmode="ISO-3",
        color="Continent",
        hover_name="Country Name",
        hover_data=["WB Region"],
        projection="natural earth",
        title=f"{title} - countries grouped by continent ({latest_year})"
    )
    fig1.show()

    print("8. Geo map by region")
    fig2 = px.choropleth(
        latest_df.dropna(subset=["WB Region"]),
        locations="Country Code",
        locationmode="ISO-3",
        color="WB Region",
        hover_name="Country Name",
        hover_data=["Continent"],
        projection="natural earth",
        title=f"{title} - countries grouped by region ({latest_year})"
    )
    fig2.show()

    print("9. Geo heatmap by value")
    map_df = latest_df.dropna(subset=[value_name]).copy()
    color_col = value_name

    if log_scale:
        map_df = map_df[map_df[value_name] > 0].copy()
        map_df["log10_value"] = np.log10(map_df[value_name])
        color_col = "log10_value"

    fig3 = px.choropleth(
        map_df,
        locations="Country Code",
        locationmode="ISO-3",
        color=color_col,
        hover_name="Country Name",
        hover_data={
            value_name: True,
            "Continent": True,
            "WB Region": True,
            "log10_value": False if "log10_value" in map_df.columns else False
        },
        color_continuous_scale="YlGnBu",
        projection="natural earth",
        title=f"{title} - geo heatmap ({latest_year})"
    )
    fig3.show()

    return raw_df, df, latest_df
```

```python
gdp_raw, gdp_long, gdp_latest = run_exploration(
    path=GDP_PATH,
    value_name="GDP per capita (current US$)",
    title="GDP per capita (current US$)",
    log_scale=True
)
```

```python
life_raw, life_long, life_latest = run_exploration(
    path=LIFE_PATH,
    value_name="Life expectancy at birth, total (years)",
    title="Life expectancy at birth, total (years)",
    log_scale=False
)
```

```python
pop_raw, pop_long, pop_latest = run_exploration(
    path=POP_PATH,
    value_name="Population, total",
    title="Population, total",
    log_scale=True
)
```

```python
CONTINENT_MAP = {
    "AF": "Africa",
    "AS": "Asia",
    "EU": "Europe",
    "NA": "North America",
    "OC": "Oceania",
    "SA": "South America",
    "AN": "Antarctica",
}

SPECIAL_CONTINENT = {
    "XKX": "Europe",        # Kosovo
    "SXM": "North America", # Sint Maarten
}

def country_to_continent(alpha3, alpha2=None):
    alpha3 = str(alpha3).strip()
    if alpha3 in SPECIAL_CONTINENT:
        return SPECIAL_CONTINENT[alpha3]

    try:
        if pd.isna(alpha2) or str(alpha2).strip() == "":
            alpha2 = pc.country_alpha3_to_country_alpha2(alpha3)
        continent_code = pc.country_alpha2_to_continent_code(str(alpha2).strip())
        return CONTINENT_MAP.get(continent_code)
    except:
        return None

def fetch_world_bank_metadata():
    url = "https://api.worldbank.org/v2/country?format=json&per_page=400"
    with urlopen(url) as response:
        payload = json.load(response)

    rows = []
    for item in payload[1]:
        rows.append({
            "Country Code": item.get("id"),
            "ISO2 Code": item.get("iso2Code"),
            "WB Country Name": item.get("name"),
            "WB Region": (item.get("region") or {}).get("value", ""),
            "Admin Region": (item.get("adminregion") or {}).get("value", ""),
            "Income Level": (item.get("incomeLevel") or {}).get("value", ""),
        })

    meta = pd.DataFrame(rows)
    meta["WB Region"] = meta["WB Region"].fillna("").str.strip().replace("", "Aggregates")
    meta["Is Country"] = meta["WB Region"].ne("Aggregates")
    meta["Continent"] = meta.apply(
        lambda row: country_to_continent(row["Country Code"], row["ISO2 Code"]),
        axis=1
    )
    return meta

metadata = fetch_world_bank_metadata()
metadata.head()
```

```python
def load_wdi(path):
    df = pd.read_csv(path, skiprows=4)
    df = df.drop(columns=["Unnamed: 69"], errors="ignore")

    year_cols = [c for c in df.columns if str(c).isdigit()]
    df[year_cols] = df[year_cols].apply(pd.to_numeric, errors="coerce")
    return df

def to_long(df, value_name):
    year_cols = [c for c in df.columns if str(c).isdigit()]
    long_df = df.melt(
        id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
        value_vars=year_cols,
        var_name="Year",
        value_name=value_name
    )
    long_df["Year"] = pd.to_numeric(long_df["Year"], errors="coerce").astype("Int64")
    long_df[value_name] = pd.to_numeric(long_df[value_name], errors="coerce")
    return long_df
```

```python
def draw_standard_maps(latest_df, value_name, title, latest_year, log_scale=False):
    value_fmt = ":,.2f"

    print("7. Geo map by continent")
    fig1 = px.choropleth(
        latest_df.dropna(subset=["Continent"]).copy(),
        locations="Country Code",
        locationmode="ISO-3",
        color="Continent",
        hover_name="Country Name",
        hover_data={
            value_name: value_fmt,
            "WB Region": True
        },
        projection="natural earth",
        title=f"{title} - countries grouped by continent ({latest_year})"
    )
    fig1.show()

    print("8. Geo map by region")
    fig2 = px.choropleth(
        latest_df.dropna(subset=["WB Region"]).copy(),
        locations="Country Code",
        locationmode="ISO-3",
        color="WB Region",
        hover_name="Country Name",
        hover_data={
            value_name: value_fmt,
            "Continent": True
        },
        projection="natural earth",
        title=f"{title} - countries grouped by region ({latest_year})"
    )
    fig2.show()

    print("9. Geo heatmap by value")
    map_df = latest_df.dropna(subset=[value_name]).copy()
    color_col = value_name

    if log_scale:
        map_df = map_df[map_df[value_name] > 0].copy()
        map_df["log10_value"] = np.log10(map_df[value_name])
        color_col = "log10_value"

    hover_dict = {
        value_name: value_fmt,
        "Continent": True,
        "WB Region": True
    }

    if "log10_value" in map_df.columns:
        hover_dict["log10_value"] = False

    fig3 = px.choropleth(
        map_df,
        locations="Country Code",
        locationmode="ISO-3",
        color=color_col,
        hover_name="Country Name",
        hover_data=hover_dict,
        color_continuous_scale="YlGnBu",
        projection="natural earth",
        title=f"{title} - geo heatmap ({latest_year})"
    )
    fig3.show()
```

```python
def draw_population_maps(latest_df, value_name, title, latest_year, log_scale=True):
    continent_totals = (
        latest_df.dropna(subset=["Continent", value_name])
        .groupby("Continent", as_index=False)[value_name]
        .sum()
        .rename(columns={value_name: "Continent Total Population"})
    )

    region_totals = (
        latest_df.dropna(subset=["WB Region", value_name])
        .groupby("WB Region", as_index=False)[value_name]
        .sum()
        .rename(columns={value_name: "Region Total Population"})
    )

    hover_df = (
        latest_df.merge(continent_totals, on="Continent", how="left")
                 .merge(region_totals, on="WB Region", how="left")
    )

    print("7. Geo map by continent")
    fig1 = px.choropleth(
        hover_df.dropna(subset=["Continent"]).copy(),
        locations="Country Code",
        locationmode="ISO-3",
        color="Continent",
        hover_name="Country Name",
        custom_data=[
            "Country Code",
            "Continent",
            "WB Region",
            value_name,
            "Continent Total Population",
        ],
        projection="natural earth",
        title=f"{title} - countries grouped by continent ({latest_year})"
    )

    fig1.update_traces(
        hovertemplate=
        "<b>%{hovertext}</b><br>"
        "Country Code: %{customdata[0]}<br>"
        "Continent: %{customdata[1]}<br>"
        "Region: %{customdata[2]}<br>"
        "Country Population: %{customdata[3]:,.0f}<br>"
        "Total Population of Continent: %{customdata[4]:,.0f}"
        "<extra></extra>"
    )
    fig1.show()

    print("8. Geo map by region")
    fig2 = px.choropleth(
        hover_df.dropna(subset=["WB Region"]).copy(),
        locations="Country Code",
        locationmode="ISO-3",
        color="WB Region",
        hover_name="Country Name",
        custom_data=[
            "Country Code",
            "Continent",
            "WB Region",
            value_name,
            "Region Total Population",
        ],
        projection="natural earth",
        title=f"{title} - countries grouped by region ({latest_year})"
    )

    fig2.update_traces(
        hovertemplate=
        "<b>%{hovertext}</b><br>"
        "Country Code: %{customdata[0]}<br>"
        "Continent: %{customdata[1]}<br>"
        "Region: %{customdata[2]}<br>"
        "Country Population: %{customdata[3]:,.0f}<br>"
        "Total Population of Region: %{customdata[4]:,.0f}"
        "<extra></extra>"
    )
    fig2.show()

    print("9. Geo heatmap by value")
    map_df = hover_df.dropna(subset=[value_name]).copy()
    color_col = value_name

    if log_scale:
        map_df = map_df[map_df[value_name] > 0].copy()
        map_df["log10_value"] = np.log10(map_df[value_name])
        color_col = "log10_value"

    hover_dict = {
        value_name: ":,.0f",
        "Continent": True,
        "WB Region": True
    }

    if "log10_value" in map_df.columns:
        hover_dict["log10_value"] = False

    fig3 = px.choropleth(
        map_df,
        locations="Country Code",
        locationmode="ISO-3",
        color=color_col,
        hover_name="Country Name",
        hover_data=hover_dict,
        color_continuous_scale="YlGnBu",
        projection="natural earth",
        title=f"{title} - geo heatmap ({latest_year})"
    )
    fig3.show()
```

```python
def run_exploration(path, value_name, title, log_scale=False, is_population=False):
    raw_df = load_wdi(path)
    long_df = to_long(raw_df, value_name)

    df = long_df.merge(
        metadata[["Country Code", "WB Region", "Admin Region", "Continent", "Is Country"]],
        on="Country Code",
        how="left"
    )

    year_cols = [c for c in raw_df.columns if str(c).isdigit()]
    latest_year = int(df.loc[df["Is Country"] & df[value_name].notna(), "Year"].max())
    latest_df = df[(df["Year"] == latest_year) & (df["Is Country"])].copy()

    summary_df = pd.DataFrame({
        "Metric": [
            "So dong",
            "So cot",
            "So cot nam",
            "So dong country/territory",
            "So dong aggregate",
            "Tong missing o cac cot nam",
            "Ty le missing o cac cot nam",
            "Nam moi nhat co du lieu cap quoc gia",
            "So nuoc co du lieu o nam moi nhat",
        ],
        "Value": [
            raw_df.shape[0],
            raw_df.shape[1],
            len(year_cols),
            int(metadata["Is Country"].sum()),
            int((~metadata["Is Country"]).sum()),
            int(raw_df[year_cols].isna().sum().sum()),
            round(raw_df[year_cols].isna().sum().sum() / raw_df[year_cols].size, 4),
            latest_year,
            int(latest_df[value_name].notna().sum()),
        ]
    })

    desc_df = latest_df[value_name].dropna().describe(
        percentiles=[0.1, 0.25, 0.5, 0.75, 0.9]
    ).to_frame(name="Value")

    continent_df = (
        latest_df.dropna(subset=["Continent", value_name])
        .groupby("Continent")[value_name]
        .agg(["count", "mean", "median", "min", "max"])
        .sort_values("count", ascending=False)
        .reset_index()
    )

    region_df = (
        latest_df.dropna(subset=["WB Region", value_name])
        .groupby("WB Region")[value_name]
        .agg(["count", "mean", "median", "min", "max"])
        .sort_values("count", ascending=False)
        .reset_index()
    )

    coverage_df = (
        df[df["Is Country"]]
        .groupby("Year")[value_name]
        .agg(non_null_count="count")
        .reset_index()
    )

    display(Markdown(f"# {title}"))
    display(Markdown(f"**Latest year with country-level data:** `{latest_year}`"))

    print("1. Basic statistics")
    display(summary_df)

    print("2. Distribution statistics at latest year")
    display(desc_df)

    print("3. Group by continent")
    display(continent_df)

    print("4. Group by region")
    display(region_df)

    print("5. Data coverage by year")
    plt.figure(figsize=(13, 4))
    sns.lineplot(data=coverage_df, x="Year", y="non_null_count", marker="o")
    plt.title(f"{title} - so nuoc co du lieu theo nam")
    plt.xlabel("Year")
    plt.ylabel("Country count")
    plt.show()

    print("6. Distribution plot")
    plot_series = latest_df[value_name].dropna()
    if log_scale:
        plot_series = plot_series[plot_series > 0]

    fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
    sns.histplot(plot_series, bins=30, kde=True, ax=axes[0], color="#2a9d8f")
    sns.boxplot(x=plot_series, ax=axes[1], color="#e9c46a")

    if log_scale:
        axes[0].set_xscale("log")
        axes[1].set_xscale("log")
        axes[0].set_xlabel(f"{value_name} (log scale)")
        axes[1].set_xlabel(f"{value_name} (log scale)")
    else:
        axes[0].set_xlabel(value_name)
        axes[1].set_xlabel(value_name)

    axes[0].set_title(f"{title} - histogram")
    axes[1].set_title(f"{title} - boxplot")
    plt.tight_layout()
    plt.show()

    if is_population:
        draw_population_maps(latest_df, value_name, title, latest_year, log_scale=log_scale)
    else:
        draw_standard_maps(latest_df, value_name, title, latest_year, log_scale=log_scale)

    return raw_df, df, latest_df
```

```python
# Figure 4.1 - Missing-data structure across core and extended indicators
# Final clean version with fixed right-panel header spacing

from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="white")
plt.rcParams["figure.figsize"] = (15, 7)
pd.options.display.float_format = "{:,.2f}".format

BASE_DIR = Path("/Users/tonytony/Final Project")
PANEL_PATH = BASE_DIR / "Data" / "Cleaned" / "panel_with_event_dummies_and_extra_drivers.csv"
OUTPUT_DIR = BASE_DIR / "output" / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

panel_df = pd.read_csv(PANEL_PATH)

indicator_map = {
    "gdp_per_capita_usd": "GDP per capita",
    "population_total": "Population",
    "life_expectancy_years": "Life expectancy",
    "inflation_pct_clean": "Inflation",
    "unemployment_pct_clean": "Unemployment",
    "internet_users_pct_clean": "Internet usage",
}

core_cols = {"gdp_per_capita_usd", "population_total", "life_expectancy_years"}

summary_rows = []
n_total = len(panel_df)

for col, label in indicator_map.items():
    non_null = int(panel_df[col].notna().sum())
    missing = int(panel_df[col].isna().sum())
    completeness_pct = non_null / n_total * 100
    missing_pct = missing / n_total * 100
    group = "Core indicators" if col in core_cols else "Extended indicators"

    summary_rows.append({
        "indicator": label,
        "group": group,
        "non_null": non_null,
        "missing": missing,
        "completeness_pct": completeness_pct,
        "missing_pct": missing_pct,
    })

summary_df = pd.DataFrame(summary_rows)
summary_df["group_order"] = summary_df["group"].map({
    "Core indicators": 0,
    "Extended indicators": 1
})
summary_df = (
    summary_df
    .sort_values(["group_order", "missing_pct", "indicator"])
    .drop(columns="group_order")
    .reset_index(drop=True)
)

display(summary_df)

plot_df = summary_df.iloc[::-1].copy()

# Colors
core_color = "#355CDE"
extended_color = "#E58E32"
core_fill = "#EEF3FF"
extended_fill = "#FFF4E8"
text_dark = "#1F2937"
text_mid = "#4B5563"
line_light = "#D8DEE8"
panel_bg = "#FCFDFE"

bar_colors = [
    core_color if g == "Core indicators" else extended_color
    for g in plot_df["group"]
]
row_fills = [
    core_fill if g == "Core indicators" else extended_fill
    for g in plot_df["group"]
]

fig = plt.figure(figsize=(15.2, 7.2), facecolor="white")
gs = fig.add_gridspec(1, 2, width_ratios=[3.8, 2.2], wspace=0.035)

ax = fig.add_subplot(gs[0, 0])
ax_tbl = fig.add_subplot(gs[0, 1], sharey=ax)

# =========================
# Left panel: missing share
# =========================
ax.set_facecolor(panel_bg)

for i, (_, row) in enumerate(plot_df.iterrows()):
    ax.axhspan(i - 0.34, i + 0.34, color=row_fills[i], zorder=0)

bars = ax.barh(
    plot_df["indicator"],
    plot_df["missing_pct"],
    color=bar_colors,
    edgecolor="none",
    height=0.56,
    zorder=3
)

for bar, value, group in zip(bars, plot_df["missing_pct"], plot_df["group"]):
    y = bar.get_y() + bar.get_height() / 2
    accent = core_color if group == "Core indicators" else extended_color

    if value <= 0.01:
        ax.text(
            1.0,
            y,
            "0.00%",
            va="center",
            ha="left",
            fontsize=10,
            color=accent,
            weight="bold",
            bbox=dict(
                boxstyle="round,pad=0.18",
                facecolor="white",
                edgecolor="none",
                alpha=0.95
            )
        )
    else:
        ax.text(
            value + 0.7,
            y,
            f"{value:.2f}%",
            va="center",
            ha="left",
            fontsize=10,
            color=accent,
            weight="bold"
        )

ax.set_xlim(0, max(56, plot_df["missing_pct"].max() + 9))
ax.set_xlabel("Missing share of observations (%)", fontsize=11, color=text_dark)
ax.set_ylabel("")
ax.set_title(
    "A. Missing-share comparison",
    loc="left",
    pad=14,
    color=text_dark,
    weight="bold"
)

ax.set_xticks([0, 10, 20, 30, 40, 50])
ax.tick_params(axis="x", colors=text_mid, labelsize=10)
ax.tick_params(axis="y", colors=text_dark, labelsize=11)

ax.grid(axis="x", linestyle="--", alpha=0.16, zorder=1)
ax.grid(axis="y", visible=False)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color(line_light)
ax.spines["bottom"].set_color(line_light)

# =========================
# Right panel: summary table
# =========================
ax_tbl.set_facecolor("white")
ax_tbl.set_xlim(0, 3.55)
ax_tbl.set_ylim(ax.get_ylim())
ax_tbl.axis("off")

x_group = 0.22
x_non_null = 1.58
x_missing = 2.43
x_complete = 3.28

# panel title placed higher
ax_tbl.text(
    0.05, 1.055,
    "B. Data-availability summary",
    transform=ax_tbl.transAxes,
    fontsize=13,
    fontweight="bold",
    color=text_dark,
    ha="left",
    va="bottom"
)

# table header placed lower to avoid overlap
header_y = len(plot_df) - 0.55
ax_tbl.text(x_group, header_y, "Type", fontsize=10.5, weight="bold", color=text_dark, ha="left")
ax_tbl.text(x_non_null, header_y, "Non-null", fontsize=10.5, weight="bold", color=text_dark, ha="right")
ax_tbl.text(x_missing, header_y, "Missing", fontsize=10.5, weight="bold", color=text_dark, ha="right")
ax_tbl.text(x_complete, header_y, "Complete %", fontsize=10.5, weight="bold", color=text_dark, ha="right")

ax_tbl.hlines(
    y=header_y - 0.20,
    xmin=0.10,
    xmax=3.38,
    color=line_light,
    linewidth=1
)

for i, (_, row) in enumerate(plot_df.iterrows()):
    y = row["indicator"]
    fill = row_fills[i]
    accent = core_color if row["group"] == "Core indicators" else extended_color
    group_short = "Core" if row["group"] == "Core indicators" else "Extended"

    ax_tbl.axhspan(i - 0.34, i + 0.34, color=fill, zorder=0)

    ax_tbl.text(x_group, y, group_short, va="center", ha="left", fontsize=10, color=text_mid)
    ax_tbl.text(x_non_null, y, f"{int(row['non_null']):,}", va="center", ha="right", fontsize=10, color=text_dark)
    ax_tbl.text(x_missing, y, f"{int(row['missing']):,}", va="center", ha="right", fontsize=10, color=text_dark)
    ax_tbl.text(
        x_complete,
        y,
        f"{row['completeness_pct']:.2f}%",
        va="center",
        ha="right",
        fontsize=10,
        color=accent,
        weight="bold"
    )

# Layout only
fig.subplots_adjust(top=0.81, left=0.10, right=0.98, bottom=0.10)

save_path = OUTPUT_DIR / "figure_4_1_missingness_structure_final_clean.png"
plt.savefig(save_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved to:", save_path)
```
