```python
unnamed_cols = [col for col in pop_df.columns if str(col).startswith("Unnamed")]

print("Technical / junk columns:")
print(unnamed_cols if unnamed_cols else "No unnamed columns found")
```

```python
SPECIAL_COUNTRY_CODES = {"XKX", "SXM"}

def is_country_code(code):
    code = str(code).strip()
    if code in SPECIAL_COUNTRY_CODES:
        return True
    try:
        alpha2 = pc.country_alpha3_to_country_alpha2(code)
        pc.country_alpha2_to_continent_code(alpha2)
        return True
    except:
        return False

country_mask = pop_df["Country Code"].astype(str).apply(is_country_code)

print("Country / territory rows:", int(country_mask.sum()))
print("Aggregate rows:", int((~country_mask).sum()))
```

## Phân biệt country rows và aggregate rows

Điều này rất quan trọng vì dataset WDI không chỉ chứa dữ liệu của từng nước, mà còn chứa:

Nếu không tách riêng, kết quả phân tích có thể bị sai lệch.

## Kiểm tra `Indicator Name` và `Indicator Code`

Cell này dùng để xác nhận dataset có đúng là chỉ chứa 1 chỉ số duy nhất hay không.

Đối với file `population.csv`, kỳ vọng là:
- `Indicator Name` chỉ có 1 giá trị là `Population, total`,
- `Indicator Code` chỉ có 1 giá trị tương ứng.

Điều này cho thấy file hiện tại là một dataset đơn biến theo thời gian, không phải file gộp nhiều chỉ số khác nhau.


```python
year_non_null = pop_df[numerical_cols].notna().sum().sort_index()

coverage_df = pd.DataFrame({
    "Year": year_non_null.index,
    "non_null_count": year_non_null.values,
    "non_null_ratio_pct": (year_non_null.values / len(pop_df) * 100).round(2)
})

coverage_df.tail(15)
```

## Kiểm tra độ đầy đủ dữ liệu theo từng năm

Cell này tính số lượng dòng có dữ liệu ở mỗi cột năm.

Output cho biết:
- năm nào có coverage cao,
- năm nào bị thiếu dữ liệu,
- tỷ lệ hoàn chỉnh dữ liệu theo từng năm.

Đây là bước quan trọng để xác định:
- năm mới nhất nào đủ tốt để dùng trong phân tích,
- mức độ đầy đủ dữ liệu qua thời gian.


```python
latest_year_with_data = coverage_df.loc[coverage_df["non_null_count"] > 0, "Year"].iloc[-1]

print("Latest year with any data:", latest_year_with_data)
print("Non-null rows in latest year:", int(pop_df[latest_year_with_data].notna().sum()))
```

## Xác định năm mới nhất còn dữ liệu

Cell này tìm năm mới nhất mà dataset vẫn còn giá trị hợp lệ.

Thông tin này rất quan trọng vì:
- không phải lúc nào cột năm cuối cùng cũng có dữ liệu đầy đủ,
- khi làm EDA hoặc vẽ bản đồ, nên chọn năm mới nhất có dữ liệu thực sự.

Kết quả của cell này sẽ được dùng làm mốc cho các bước phân tích tiếp theo.


```python
issues = []

if unnamed_cols:
    issues.append(f"Co cot rac/technical column: {unnamed_cols}")

if (~country_mask).sum() > 0:
    issues.append(
        f"Dataset chua ca aggregate rows, khong chi country rows: {(~country_mask).sum()} dong aggregate, {country_mask.sum()} dong country/territory."
    )

if pop_df["Indicator Name"].nunique() == 1 and pop_df["Indicator Code"].nunique() == 1:
    issues.append("File chi chua 1 indicator duy nhat la Population, total.")

if pop_df.duplicated().sum() == 0:
    issues.append("Khong co duplicate row hoan toan trung nhau.")
else:
    issues.append(f"Co {pop_df.duplicated().sum()} duplicate rows.")

all_null_cols = pop_null.loc[pop_null["null_count"] == len(pop_df), "column"].tolist()
if all_null_cols:
    issues.append(f"Co cot nam bi null 100%: {all_null_cols}")
else:
    issues.append("Khong co cot nao bi null 100%.")

high_missing_cols = pop_null.loc[
    (pop_null["column"].isin(numerical_cols)) & (pop_null["null_ratio_pct"] >= 30),
    "column"
].tolist()

if high_missing_cols:
    issues.append(f"Cac nam co missing rat cao (>=30%): {high_missing_cols}")
else:
    issues.append("Khong co cot nam nao missing tu 30% tro len.")

count_2024 = int(pop_df["2024"].notna().sum())
issues.append(f"Cot 2024 gan nhu day du, co {count_2024}/{len(pop_df)} dong co du lieu.")

latest_series = pd.to_numeric(pop_df[latest_year_with_data], errors="coerce").dropna()
mean_val = latest_series.mean()
median_val = latest_series.median()

if median_val != 0 and mean_val / median_val >= 2:
    issues.append(
        f"Phan phoi population lech phai rat manh/outlier cao vi mean ({mean_val:,.2f}) lon hon median ({median_val:,.2f}) rat nhieu."
    )

issues.append("Du lieu dang o wide format theo nam, can melt sang long format neu muon phan tich time series hoac merge voi dataset khac.")

for i, issue in enumerate(issues, start=1):
    print(f"{i}. {issue}")
```
