```python
import sys
import subprocess

APP_PATH = "/Users/tonytony/Final Project/app.py"
LOG_PATH = "/tmp/streamlit_population_dashboard.log"

proc = subprocess.Popen(
    [
        sys.executable,
        "-m",
        "streamlit",
        "run",
        APP_PATH,
        "--server.headless",
        "true",
        "--server.port",
        "8501",
    ],
    stdout=open(LOG_PATH, "w"),
    stderr=subprocess.STDOUT,
    start_new_session=True,
)

print("Started Streamlit")
print("PID:", proc.pid)
print("Open: http://localhost:8501")
print("Log file:", LOG_PATH)
```

```python
import sys
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "pip", "install", "streamlit", "plotly"],
    capture_output=True,
    text=True
)

print("Return code:", result.returncode)
print("\nSTDOUT:\n", result.stdout[-4000:])
print("\nSTDERR:\n", result.stderr[-4000:])
```
