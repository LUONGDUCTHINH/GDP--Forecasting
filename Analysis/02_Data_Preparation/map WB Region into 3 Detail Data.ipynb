```python
import pandas as pd

GDP_PATH = "/Users/tonytony/Final Project/Data/Raw/gdp.csv"
LIFE_PATH = "/Users/tonytony/Final Project/Data/Raw/life_expectancy.csv"
POP_PATH = "/Users/tonytony/Final Project/Data/Raw/population.csv"
META_PATH = "/Users/tonytony/Final Project/Data/Raw/world_bank_country_metadata.csv"
print("GDP:", GDP_PATH)
print("Life:", LIFE_PATH)
print("Population:", POP_PATH)
print("Metadata:", META_PATH)
```

```python
wb_meta = pd.read_csv(META_PATH)

print("Metadata shape:", wb_meta.shape)
print("Metadata columns:")
print(wb_meta.columns.tolist())

wb_meta.head()
```

```python
wb_region_map = wb_meta[[
    "Country Code",
    "World Bank Region",
    "Admin Region",
    "Income Level",
    "Lending Type"
]].drop_duplicates(subset=["Country Code"]).copy()

wb_region_map = wb_region_map.rename(columns={
    "World Bank Region": "WB Region"
})

wb_region_map["Country Type"] = wb_region_map["WB Region"].apply(
    lambda x: "Aggregate" if str(x).strip() == "Aggregates" else "Country/Territory"
)

print("WB region mapping table shape:", wb_region_map.shape)
wb_region_map.head(10)
```

```python
gdp_df = pd.read_csv(GDP_PATH, skiprows=4)

print("GDP shape:", gdp_df.shape)
gdp_df.head()
```

```python
gdp_with_region = gdp_df.merge(
    wb_region_map,
    on="Country Code",
    how="left"
)

print("GDP with WB Region shape:", gdp_with_region.shape)
gdp_with_region[[
    "Country Name",
    "Country Code",
    "WB Region",
    "Admin Region",
    "Income Level",
    "Country Type"
]].head(15)
```

```python
gdp_unmapped = gdp_with_region[gdp_with_region["WB Region"].isna()][[
    "Country Name",
    "Country Code"
]]

print("Number of unmapped rows in GDP:", gdp_unmapped.shape[0])
gdp_unmapped.drop_duplicates().head(20)
```

```python

life_df = pd.read_csv(LIFE_PATH, skiprows=4)

print("Life Expectancy shape:", life_df.shape)
life_df.head()
```

```python
life_with_region = life_df.merge(
    wb_region_map,
    on="Country Code",
    how="left"
)

print("Life Expectancy with WB Region shape:", life_with_region.shape)
life_with_region[[
    "Country Name",
    "Country Code",
    "WB Region",
    "Admin Region",
    "Income Level",
    "Country Type"
]].head(15)
```

```python
life_unmapped = life_with_region[life_with_region["WB Region"].isna()][[
    "Country Name",
    "Country Code"
]]

print("Number of unmapped rows in Life Expectancy:", life_unmapped.shape[0])
life_unmapped.drop_duplicates().head(20)
```

```python
pop_df = pd.read_csv(POP_PATH, skiprows=4)

print("Population shape:", pop_df.shape)
pop_df.head()
```

```python
pop_with_region = pop_df.merge(
    wb_region_map,
    on="Country Code",
    how="left"
)

print("Population with WB Region shape:", pop_with_region.shape)
pop_with_region[[
    "Country Name",
    "Country Code",
    "WB Region",
    "Admin Region",
    "Income Level",
    "Country Type"
]].head(15)
```

```python
pop_unmapped = pop_with_region[pop_with_region["WB Region"].isna()][[
    "Country Name",
    "Country Code"
]]

print("Number of unmapped rows in Population:", pop_unmapped.shape[0])
pop_unmapped.drop_duplicates().head(20)
```

```python
print("GDP - WB Region distribution")
display(gdp_with_region["WB Region"].value_counts(dropna=False).to_frame("count"))

print("Life Expectancy - WB Region distribution")
display(life_with_region["WB Region"].value_counts(dropna=False).to_frame("count"))

print("Population - WB Region distribution")
display(pop_with_region["WB Region"].value_counts(dropna=False).to_frame("count"))
```

```python
gdp_country_only = gdp_with_region[gdp_with_region["Country Type"] == "Country/Territory"].copy()
life_country_only = life_with_region[life_with_region["Country Type"] == "Country/Territory"].copy()
pop_country_only = pop_with_region[pop_with_region["Country Type"] == "Country/Territory"].copy()

print("GDP country-only shape:", gdp_country_only.shape)
print("Life country-only shape:", life_country_only.shape)
print("Population country-only shape:", pop_country_only.shape)
```

```python
gdp_with_region.to_csv("/Users/tonytony/Final Project/Data/Raw/gdp_with_wb_region.csv", index=False)
life_with_region.to_csv("/Users/tonytony/Final Project/Data/Raw/life_with_wb_region.csv", index=False)
pop_with_region.to_csv("/Users/tonytony/Final Project/Data/Raw/pop_with_wb_region.csv", index=False)

print("Saved:")
print("gdp_with_wb_region.csv")
print("life_with_wb_region.csv")
print("pop_with_wb_region.csv")
```
